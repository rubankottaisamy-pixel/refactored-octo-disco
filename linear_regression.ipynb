```python
df_iris['iris_type'] = model.target
x_iris = df_iris.drop('iris_type', axis=1)
y_iris = df_iris['iris_type']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x_iris,y_iris,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
print(mean_squared_error(y_test,y_pred))
```

```python
print(mean_absolute_error(y_test,y_pred))
```

```python
from sklearn.metrics import mean_squared_error,r2_score
print(mean_squared_error(y_test,y_pred))
```

```python
print(mean_absolute_error(y_pred,y_test))
```

```python
r2_score(y_test,y_pred)
```

```python
plt.figure(figsize=(10,5))
plt.scatter(y_test,y_pred)
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()
```

```python
residuals = y_test - y_pred
plt.figure(figsize=(10,5))
sns.histplot(residuals,kde=True)
plt.title("Residual Distribution")
plt.show()
```

```python
sns.regplot(x=y_test,y=y_pred,ci=None,color='r')
plt.xlabel("actual")
plt.ylabel("predicted")
plt.title("regression line")
plt.show()
```
